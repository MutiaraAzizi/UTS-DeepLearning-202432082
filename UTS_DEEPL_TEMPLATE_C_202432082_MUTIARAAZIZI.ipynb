  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>
</div>

```python
#202432082_MutiaraAzizi
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432082_MutiaraAzizi
# Baca file CSV dan ubah kolom Date menjadi format tanggal
df = pd.read_csv('daily_gold_rate.csv')
df['Date'] = pd.to_datetime(df['Date'])

# Tampilkan ukuran data, 5 baris awal, tipe data, dan jumlah nilai kosong
print('Jumlah baris dan kolom:', df.shape)
print(df.head())
print(df.dtypes)
print(df.isna().sum())

# Grafik harga emas (USD) dari waktu ke waktu
plt.figure(figsize=(12, 4))
plt.plot(df['Date'], df['USD'], color='#114F6A')
plt.title('Harga Emas Harian (USD), 1985 sampai 2023')
plt.xlabel('Tanggal')
plt.ylabel('Harga (USD)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

#202432082_MutiaraAzizi
# Ambil kolom USD sebagai target, bentuknya (jumlah_baris, 1)
data = df[['USD']].values

# Bagi data 80% latih dan 20% uji, urutan waktu tidak diacak
split = int(len(data) * 0.8)
train_data = data[:split]
test_data = data[split:]

# Normalisasi ke rentang 0 sampai 1, scaler hanya belajar dari data latih
scaler = MinMaxScaler()
train_scaled = scaler.fit_transform(train_data)
test_scaled = scaler.transform(test_data)

# Sliding window: 30 hari terakhir jadi input, hari ke-31 jadi target
def create_sequences(data, window):
    X, y = [], []
    for i in range(len(data) - window):
        X.append(data[i:i + window])
        y.append(data[i + window])
    return np.array(X), np.array(y)

window = 30
X_train, y_train = create_sequences(train_scaled, window)
X_test, y_test = create_sequences(test_scaled, window)

# Ubah ke tensor PyTorch bertipe float32
X_train = torch.tensor(X_train, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.float32)
X_test = torch.tensor(X_test, dtype=torch.float32)
y_test = torch.tensor(y_test, dtype=torch.float32)

print(X_train.shape, y_train.shape, X_test.shape, y_test.shape)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432082_MutiaraAzizi
class SequenceModel(nn.Module):
    def __init__(self, input_size=1, hidden_size=64, num_layers=2):
        super(SequenceModel, self).__init__()
        # Lapisan LSTM: 1 fitur input, 64 unit tersembunyi, 2 lapis, dropout 20%
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers,
                            batch_first=True, dropout=0.2)
        # Lapisan linear mengubah 64 nilai menjadi 1 prediksi harga
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, _ = self.lstm(x)        # out berukuran (batch, 30, 64)
        out = out[:, -1, :]          # ambil hasil langkah waktu terakhir
        return self.fc(out)          # prediksi berukuran (batch, 1)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432082_MutiaraAzizi
# Hyperparameter
epochs = 50
learning_rate = 0.001
batch_size = 64

# Pakai GPU kalau ada, kalau tidak pakai CPU
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

# Objek model, fungsi loss, dan optimizer
model = SequenceModel().to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# DataLoader memecah data latih menjadi batch kecil
train_loader = DataLoader(TensorDataset(X_train, y_train),
                          batch_size=batch_size, shuffle=True)

# Daftar untuk menyimpan loss tiap epoch
train_losses = []

# Loop training
for epoch in range(epochs):
    model.train()
    total_loss = 0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()            # kosongkan gradien lama
        pred = model(xb)                 # forward pass
        loss = criterion(pred, yb)       # hitung MSE
        loss.backward()                  # backpropagation
        optimizer.step()                 # perbarui bobot
        total_loss += loss.item() * xb.size(0)
    epoch_loss = total_loss / len(train_loader.dataset)
    train_losses.append(epoch_loss)
    if (epoch + 1) % 5 == 0:
        print(f'Epoch {epoch+1}/{epochs}, Loss: {epoch_loss:.6f}')
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432082_MutiaraAzizi
plt.figure(figsize=(10, 4))
plt.plot(train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()

#202432082_MutiaraAzizi
# Prediksi pada data uji
model.eval()
with torch.no_grad():
    pred = model(X_test.to(device)).cpu().numpy()

# Kembalikan skala ke harga asli (USD)
pred_inv = scaler.inverse_transform(pred)
actual_inv = scaler.inverse_transform(y_test.numpy())

# Metrik evaluasi
rmse = np.sqrt(np.mean((pred_inv - actual_inv) ** 2))
mae = np.mean(np.abs(pred_inv - actual_inv))
mape = np.mean(np.abs((actual_inv - pred_inv) / actual_inv)) * 100
print(f'RMSE: {rmse:.2f} | MAE: {mae:.2f} | MAPE: {mape:.2f}%')

# Grafik aktual vs prediksi
test_dates = df['Date'].values[split + window:]
plt.figure(figsize=(12, 5))
plt.plot(test_dates, actual_inv, color='#114F6A', label='Data Aktual')
plt.plot(test_dates, pred_inv, color='#1FA4B1', label='Prediksi LSTM')
plt.title('Perbandingan Harga Emas Aktual vs Prediksi (USD)')
plt.xlabel('Tanggal')
plt.ylabel('Harga Emas (USD)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
